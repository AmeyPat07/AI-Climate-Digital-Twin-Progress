### Processing Methodology

The original IMD 1° × 1° maximum-temperature data is provided
as binary GRD files.

The processing workflow consists of:

1. Reading the IMD binary GRD data.
2. Decoding the temperature values.
3. Associating values with their spatial grid coordinates.
4. Filtering the grid locations corresponding to Maharashtra.
5. Converting the extracted observations into a structured
   tabular dataset.
6. Saving the processed climate data as CSV for further
   exploratory analysis and machine-learning applications.
