```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries loaded successfully.")
```

```python
df = pd.read_csv("../data/processed/maharashtra_max_temp_2015_2024.csv")

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
df.select_dtypes(include="number").iloc[:, 0].plot(
    figsize=(10, 4),
    title="Temperature Data Overview"
)

plt.xlabel("Observation")
plt.ylabel("Maximum Temperature")
plt.show()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df["date"] = pd.to_datetime(df["date"])

print("Start date:", df["date"].min())
print("End date:", df["date"].max())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df["max_temp_C"].dropna(), bins=30)

plt.title("Distribution of Maximum Daily Temperature")
plt.xlabel("Maximum Temperature (°C)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["date"], df["max_temp_C"])

plt.title("Daily Maximum Temperature")
plt.xlabel("Date")
plt.ylabel("Maximum Temperature (°C)")

plt.show()
```
